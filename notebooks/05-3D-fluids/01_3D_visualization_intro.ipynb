````python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm
%matplotlib inline
````

## 1. Creating 3-D Arrays

````python
# Create a 3-D grid
nx, ny, nz = 20, 20, 20
x = np.linspace(-1, 1, nx)
y = np.linspace(-1, 1, ny)
z = np.linspace(-1, 1, nz)

X, Y, Z = np.meshgrid(x, y, z, indexing='ij')

# Create a 3-D temperature field (example: sphere with hot center)
R = np.sqrt(X**2 + Y**2 + Z**2)
T = 100 * np.exp(-5 * R**2)  # Temperature decreases from center

print(f"3-D array shape: {T.shape}")
print(f"Temperature range: {T.min():.2f} to {T.max():.2f} °C")
````

## 2. 3-D Scatter Plot

````python
# Create a 3-D scatter plot for points above threshold
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Select points where temperature is high
threshold = 50
mask = T > threshold

scatter = ax.scatter(X[mask], Y[mask], Z[mask], 
                     c=T[mask], cmap='hot', s=20, alpha=0.6)

ax.set_xlabel('x (m)')
ax.set_ylabel('y (m)')
ax.set_zlabel('z (m)')
ax.set_title('3-D Temperature Field (T > 50°C)')
fig.colorbar(scatter, ax=ax, label='Temperature (°C)', shrink=0.5)

plt.show()
````

## 3. Slice Visualization

Visualize 2-D slices through the 3-D domain

````python
# Create slices at different locations
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# XY plane (z = 0)
z_idx = nz // 2
im1 = axes[0].contourf(x, y, T[:, :, z_idx].T, levels=20, cmap='hot')
axes[0].set_xlabel('x (m)')
axes[0].set_ylabel('y (m)')
axes[0].set_title(f'XY Plane at z = {z[z_idx]:.2f} m')
axes[0].set_aspect('equal')
plt.colorbar(im1, ax=axes[0], label='T (°C)')

# XZ plane (y = 0)
y_idx = ny // 2
im2 = axes[1].contourf(x, z, T[:, y_idx, :].T, levels=20, cmap='hot')
axes[1].set_xlabel('x (m)')
axes[1].set_ylabel('z (m)')
axes[1].set_title(f'XZ Plane at y = {y[y_idx]:.2f} m')
axes[1].set_aspect('equal')
plt.colorbar(im2, ax=axes[1], label='T (°C)')

# YZ plane (x = 0)
x_idx = nx // 2
im3 = axes[2].contourf(y, z, T[x_idx, :, :].T, levels=20, cmap='hot')
axes[2].set_xlabel('y (m)')
axes[2].set_ylabel('z (m)')
axes[2].set_title(f'YZ Plane at x = {x[x_idx]:.2f} m')
axes[2].set_aspect('equal')
plt.colorbar(im3, ax=axes[2], label='T (°C)')

plt.tight_layout()
plt.show()
````

## 4. 3-D Surface Plot

````python
# Plot isothermal surface
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# Create a surface at a specific temperature
iso_value = 50
mask = np.abs(T - iso_value) < 2  # Points near isosurface

ax.scatter(X[mask], Y[mask], Z[mask], 
           c='red', s=10, alpha=0.5)

ax.set_xlabel('x (m)')
ax.set_ylabel('y (m)')
ax.set_zlabel('z (m)')
ax.set_title(f'Isothermal Surface at T = {iso_value}°C')

plt.show()
````

## 5. 3-D Vector Field

Visualize velocity or heat flux vectors

````python
# Create a simple vector field (e.g., velocity)
# For visualization, use a coarser grid
nx_vec, ny_vec, nz_vec = 8, 8, 8
x_vec = np.linspace(-1, 1, nx_vec)
y_vec = np.linspace(-1, 1, ny_vec)
z_vec = np.linspace(-1, 1, nz_vec)

X_vec, Y_vec, Z_vec = np.meshgrid(x_vec, y_vec, z_vec, indexing='ij')

# Example: circular flow pattern
U = -Y_vec
V = X_vec
W = Z_vec * 0.5

# Plot
fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

ax.quiver(X_vec, Y_vec, Z_vec, U, V, W, 
          length=0.2, normalize=True, alpha=0.7)

ax.set_xlabel('x (m)')
ax.set_ylabel('y (m)')
ax.set_zlabel('z (m)')
ax.set_title('3-D Vector Field Example')

plt.show()
````
