```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# Set plot style
plt.style.use('seaborn-v0_8-darkgrid')
print("Libraries imported successfully!")
```

## 1. NumPy Arrays

NumPy arrays are the foundation of scientific computing in Python.

```python
# Create 1-D array
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create 2-D array (matrix)
matrix = np.zeros((5, 5))
print("1-D array shape:", x.shape)
print("2-D array shape:", matrix.shape)
```

## 2. Basic Plotting

```python
# Simple plot
plt.figure(figsize=(10, 6))
plt.plot(x, y, 'b-', linewidth=2, label='sin(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Basic Plot Example')
plt.legend()
plt.grid(True)
plt.show()
```

## 3. 2-D Arrays and Visualization

This is essential for visualizing temperature and velocity fields.

```python
# Create 2-D grid
x_2d = np.linspace(-2, 2, 50)
y_2d = np.linspace(-2, 2, 50)
X, Y = np.meshgrid(x_2d, y_2d)

# Calculate a 2-D function
Z = np.exp(-(X**2 + Y**2))

# Visualize with contour plot
plt.figure(figsize=(10, 8))
contour = plt.contourf(X, Y, Z, levels=20, cmap='viridis')
plt.colorbar(contour, label='Value')
plt.xlabel('x')
plt.ylabel('y')
plt.title('2-D Gaussian Distribution')
plt.axis('equal')
plt.show()
```
