Choose **ONE** of the following real-world systems to model:

### Option A: Shell-and-Tube Heat Exchanger

**Application:** Heat recovery in chemical processing plants

**Description:** Hot fluid flows through the tube while cold fluid flows in the shell (annular region). Model the developing velocity and temperature profiles in the tube with convective heat transfer through the wall.

**Key Physics:**
- Laminar/turbulent flow in cylindrical geometry
- Convective heat transfer with entrance effects
- Conjugate heat transfer through tube wall
- Reynolds, Prandtl, and Nusselt number analysis

**Design Questions:**
- What tube length is required to achieve desired heat transfer?
- How does flow rate affect heat exchanger efficiency?
- What material and wall thickness optimize performance?

---

### Option B: Microfluidic Mixing Device

**Application:** Lab-on-a-chip for biological assays or chemical synthesis

**Description:** Two fluid streams enter a 2D microchannel at different temperatures. Model the velocity field and thermal mixing as streams combine and flow downstream.

**Key Physics:**
- Low Reynolds number flow (Stokes regime)
- Diffusion-dominated mixing vs. convective transport
- Peclet number analysis
- Entrance effects and developing flow

**Design Questions:**
- What channel geometry promotes fastest mixing?
- How does flow rate affect mixing length?
- Can you optimize for both residence time and mixing efficiency?

---

### Option C: Stirred Bioreactor with Cooling Jacket

**Application:** Production of recombinant proteins or industrial fermentation

**Description:** Model a simplified 2D cylindrical bioreactor with internal heat generation (metabolic activity) and external cooling through the vessel wall. Include natural convection effects.

**Key Physics:**
- Natural convection driven by temperature gradients
- Volumetric heat generation
- Rayleigh number and convection patterns
- Transient approach to steady state

**Design Questions:**
- Can cooling jacket maintain target temperature range?
- How do hot spots form and can they be minimized?
- What cooling water flow rate is required?

---

### Option D: Packed Bed Catalytic Reactor

**Application:** Catalytic conversion in petrochemical processing

**Description:** Model flow through a 2D porous packed bed with exothermic reaction heat generation. Treat the bed as a porous medium with effective properties.

**Key Physics:**
- Flow through porous media (Darcy's law or Brinkman equation)
- Heat generation with temperature-dependent reaction rate
- Effective thermal conductivity and dispersion
- Potential for thermal runaway

**Design Questions:**
- What inlet conditions prevent thermal runaway?
- How does bed height affect conversion and temperature?
- Where should temperature sensors be placed?

---

### Option E: Blood Flow in Arterial Bifurcation

**Application:** Cardiovascular device design or disease modeling

**Description:** Model blood flow and heat transfer in a 2D arterial segment with a bifurcation (Y-junction). Include viscosity effects and metabolic heat transfer at walls.

**Key Physics:**
- Non-Newtonian fluid behavior (power-law or Carreau model)
- Pulsatile vs. steady flow
- Wall shear stress distribution (relevant for atherosclerosis)
- Coupled heat transfer with surrounding tissue

**Design Questions:**
- Where do low shear stress regions occur (disease prone)?
- How does bifurcation angle affect flow distribution?
- What is the temperature distribution in hypothermia/hyperthermia?
